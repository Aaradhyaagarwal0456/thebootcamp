```python
# Importing the libraries used for image processing and displaying images

import cv2
import numpy as np
import matplotlib.pyplot as plt
from PIL import Image
import os

print("All libraries loaded successfully!")
```

```python
# Load the image from the project folder
# OpenCV reads images in BGR format by default

img = cv2.imread("GitHub_CV_Image.jpg")

# Check whether the image was loaded correctly
if img is None:
    print("Error: Image not found. Please check the file name and location.")
else:
    print("Image loaded successfully!")
    print("Image type:", type(img))
    print("Image shape:", img.shape)
```

```python
# Convert BGR to RGB because Matplotlib expects RGB format

img_rgb = cv2.cvtColor(img, cv2.COLOR_BGR2RGB)

# Display the original image

plt.figure(figsize=(7, 7))
plt.imshow(img_rgb)
plt.title("Original Image")
plt.axis("off")
plt.show()
```

```python
# Change the image dimensions to 300 x 300 pixels

img_resize = cv2.resize(img, (300, 300))

# Display the resized image

plt.figure(figsize=(6, 6))
plt.imshow(cv2.cvtColor(img_resize, cv2.COLOR_BGR2RGB))
plt.title("Resized Image")
plt.axis("off")
plt.show()

print("Original Shape:", img.shape)
print("Resized Shape :", img_resize.shape)
```

```python
# Flip the image horizontally
# flipCode = 1 means horizontal flipping

img_flip_horizontal = cv2.flip(img, 1)

plt.figure(figsize=(7, 7))
plt.imshow(cv2.cvtColor(img_flip_horizontal, cv2.COLOR_BGR2RGB))
plt.title("Horizontal Flip")
plt.axis("off")
plt.show()
```

```python
# Flip the image vertically
# flipCode = 0 means vertical flipping

img_flip_vertical = cv2.flip(img, 0)

plt.figure(figsize=(7, 7))
plt.imshow(cv2.cvtColor(img_flip_vertical, cv2.COLOR_BGR2RGB))
plt.title("Vertical Flip")
plt.axis("off")
plt.show()
```

```python
# Rotate the image by 90 degrees clockwise

img_rotate = cv2.rotate(img, cv2.ROTATE_90_CLOCKWISE)

plt.figure(figsize=(7, 7))
plt.imshow(cv2.cvtColor(img_rotate, cv2.COLOR_BGR2RGB))
plt.title("Rotated Image - 90°")
plt.axis("off")
plt.show()
```

```python
# Crop a selected portion of the image
# Format: image[height_start:height_end, width_start:width_end]

img_crop = img[100:380, 100:380]

plt.figure(figsize=(6, 6))
plt.imshow(cv2.cvtColor(img_crop, cv2.COLOR_BGR2RGB))
plt.title("Cropped Image")
plt.axis("off")
plt.show()
```

```python
# Apply Gaussian Blur to make the image look softer

img_blur = cv2.GaussianBlur(img, (15, 15), 0)

plt.figure(figsize=(7, 7))
plt.imshow(cv2.cvtColor(img_blur, cv2.COLOR_BGR2RGB))
plt.title("Blurred Image")
plt.axis("off")
plt.show()
```

```python
# Sharpening helps make edges and details more noticeable

kernel = np.array([
    [0, -1, 0],
    [-1, 5, -1],
    [0, -1, 0]
])

img_sharp = cv2.filter2D(img, -1, kernel)

plt.figure(figsize=(7, 7))
plt.imshow(cv2.cvtColor(img_sharp, cv2.COLOR_BGR2RGB))
plt.title("Sharpened Image")
plt.axis("off")
plt.show()
```

```python
# Increase the contrast to make light and dark areas more distinct

contrast = 1.5

img_contrast = cv2.convertScaleAbs(
    img,
    alpha=contrast,
    beta=0
)

plt.figure(figsize=(7, 7))
plt.imshow(cv2.cvtColor(img_contrast, cv2.COLOR_BGR2RGB))
plt.title("High Contrast Image")
plt.axis("off")
plt.show()
```

```python
# Increase the brightness of the image
# beta controls the brightness level

brightness = 60

img_bright = cv2.convertScaleAbs(
    img,
    alpha=1.0,
    beta=brightness
)

plt.figure(figsize=(7, 7))
plt.imshow(cv2.cvtColor(img_bright, cv2.COLOR_BGR2RGB))
plt.title("Brightened Image")
plt.axis("off")
plt.show()
```

```python
# Detect the major edges and boundaries present in the image

edges = cv2.Canny(img, 100, 200)

plt.figure(figsize=(7, 7))
plt.imshow(edges, cmap="gray")
plt.title("Edge Detection")
plt.axis("off")
plt.show()
```

```python
# Save the processed images so they can be used later
# or uploaded along with the project on GitHub

cv2.imwrite("Photoshop_Resized.jpg", img_resize)
cv2.imwrite("Photoshop_Flip_Horizontal.jpg", img_flip_horizontal)
cv2.imwrite("Photoshop_Flip_Vertical.jpg", img_flip_vertical)
cv2.imwrite("Photoshop_Rotated.jpg", img_rotate)
cv2.imwrite("Photoshop_Cropped.jpg", img_crop)
cv2.imwrite("Photoshop_Blur.jpg", img_blur)
cv2.imwrite("Photoshop_Sharpened.jpg", img_sharp)
cv2.imwrite("Photoshop_Brightened.jpg", img_bright)
cv2.imwrite("Photoshop_Contrast.jpg", img_contrast)
cv2.imwrite("Photoshop_Edges.jpg", edges)

print("All edited images saved successfully!")
```

```python
# Compare the original image with one of the edited versions

plt.figure(figsize=(12, 6))

plt.subplot(1, 2, 1)
plt.imshow(cv2.cvtColor(img, cv2.COLOR_BGR2RGB))
plt.title("Original Image")
plt.axis("off")

plt.subplot(1, 2, 2)
plt.imshow(cv2.cvtColor(img_sharp, cv2.COLOR_BGR2RGB))
plt.title("After Sharpening")
plt.axis("off")

plt.tight_layout()
plt.show()
```

```python
# A single function to perform different image editing operations

def photoshop(image, operation):

    if operation == "resize":
        return cv2.resize(image, (300, 300))

    elif operation == "horizontal_flip":
        return cv2.flip(image, 1)

    elif operation == "vertical_flip":
        return cv2.flip(image, 0)

    elif operation == "rotate":
        return cv2.rotate(image, cv2.ROTATE_90_CLOCKWISE)

    elif operation == "grayscale":
        return cv2.cvtColor(image, cv2.COLOR_BGR2GRAY)

    elif operation == "blur":
        return cv2.GaussianBlur(image, (15, 15), 0)

    elif operation == "sharpen":
        kernel = np.array([
            [0, -1, 0],
            [-1, 5, -1],
            [0, -1, 0]
        ])
        return cv2.filter2D(image, -1, kernel)

    elif operation == "bright":
        return cv2.convertScaleAbs(image, alpha=1.0, beta=60)

    elif operation == "contrast":
        return cv2.convertScaleAbs(image, alpha=1.5, beta=0)

    elif operation == "edges":
        return cv2.Canny(image, 100, 200)

    else:
        print("Unknown operation!")
        return image
```

```python
# Choose the operation we want to apply

result = photoshop(img, "grayscale")

# Display the result

plt.figure(figsize=(7, 7))

if len(result.shape) == 2:
    plt.imshow(result, cmap="gray")
else:
    plt.imshow(cv2.cvtColor(result, cv2.COLOR_BGR2RGB))

plt.title("PhotoShop Result")
plt.axis("off")
plt.show()
```
