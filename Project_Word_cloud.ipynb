```python
import nltk
from nltk.corpus import stopwords
from nltk.tokenize import word_tokenize
from wordcloud import WordCloud
import matplotlib.pyplot as plt
from collections import Counter
```

```python
nltk.download('punkt')
nltk.download('punkt_tab')
nltk.download('stopwords')
```

```python
text = """
Photography photography photography photography photography
Photography travelling travelling travelling travelling travelling
Coding coding coding coding coding coding
Cricket cricket cricket cricket cricket
Sketching sketching sketching sketching
Gaming gaming gaming gaming
Technology technology technology
Cooking cooking cooking
Fitness fitness fitness
Cycling cycling cycling
Chess chess chess
Music music music
Art art art
Programming programming
Python python
Filmmaking filmmaking
Gym gym
Gardening gardening
Stargazing stargazing
Astronomy astronomy
Editing editing
Designing designing
Trekking trekking
Calligraphy calligraphy
Journaling journaling
Badminton badminton
Football football
Painting painting
Guitar guitar
Singing singing
Swimming swimming
Hiking hiking
Podcasting podcasting
Blogging blogging
Robotics robotics
Reading reading
Writing writing
Machine learning machine learning
Artificial intelligence artificial intelligence
Nature nature
Travel travel
Photography photography
Coding coding
Cricket cricket
Travelling travelling
Sketching sketching
Gaming gaming
Technology technology
Cooking cooking
Fitness fitness
Photography coding cricket travelling
"""
print(text)
```

```python
words = word_tokenize(text.lower())

print(words)
```

```python
stop_words = set(stopwords.words('english'))

filtered_words = [
    word for word in words
    if word.isalpha() and word not in stop_words
]

print(filtered_words)
```

```python
word_freq = Counter(filtered_words)

print("Hobby Frequencies:\n")

for word, frequency in word_freq.most_common():
    print(word, ":", frequency)
```

```python
wordcloud = WordCloud(
    width=1000,
    height=500,
    background_color="white",
    colormap="Dark2",
    max_words=40,
    min_font_size=10,
    max_font_size=90,
    margin=5,
    random_state=42
).generate_from_frequencies(word_freq)

plt.figure(figsize=(14, 7))

plt.imshow(wordcloud, interpolation="bilinear")

plt.axis("off")

plt.show()
```

```python
# Create Word Cloud with a clean center space

import numpy as np

width = 1000
height = 500

# Create mask
# Black = words allowed
# White = reserved/empty space

mask = np.zeros((height, width), dtype=np.uint8)

# Make the whole area available for words
mask[:, :] = 0

# Reserve ONLY the center for AARADHYA HOBBIES
mask[210:290, 330:670] = 255

wordcloud = WordCloud(
    width=width,
    height=height,
    background_color="white",
    mask=mask,
    colormap="Dark2",
    max_words=40,
    min_font_size=10,
    max_font_size=85,
    margin=6,
    random_state=42,
    font_path="C:/Windows/Fonts/arialbd.ttf"
).generate_from_frequencies(word_freq)

plt.figure(figsize=(14, 7))

plt.imshow(wordcloud, interpolation="bilinear")

plt.axis("off")

plt.show()
```

```python
# ============================================
# FINAL HEADING + SINGLE BORDER
# ============================================

from PIL import ImageDraw, ImageFont

# Convert current word cloud to image
image = wordcloud.to_image().convert("RGB")

draw = ImageDraw.Draw(image)


# ------------------------------------------------
# AARADHYA HOBBIES - DISTINCT BOLD FONT
# ------------------------------------------------

# Impact gives a clearly different bold look
font_path = "C:/Windows/Fonts/impact.ttf"

heading_font = ImageFont.truetype(
    font_path,
    36
)

heading = "AARADHYA HOBBIES"


# Find heading dimensions
bbox = draw.textbbox(
    (0, 0),
    heading,
    font=heading_font
)

text_width = bbox[2] - bbox[0]
text_height = bbox[3] - bbox[1]


# Center of image
x = (1000 - text_width) // 2
y = (500 - text_height) // 2


# ------------------------------------------------
# CLEAN CENTER BACKGROUND
# ------------------------------------------------

draw.rounded_rectangle(
    (
        x - 14,
        y - 7,
        x + text_width + 14,
        y + text_height + 7
    ),
    radius=8,
    fill="white"
)


# ------------------------------------------------
# HEADING
# ------------------------------------------------

draw.text(
    (x, y),
    heading,
    fill="#32105F",
    font=heading_font
)


# ------------------------------------------------
# SINGLE CLEAN BORDER
# ------------------------------------------------

draw.rounded_rectangle(
    (8, 8, 992, 492),
    radius=22,
    outline="#32105F",
    width=4
)


# ------------------------------------------------
# SMALL CORNER ACCENTS
# ------------------------------------------------

# Four tiny diamonds - part of the same border design

corner_points = [
    (25, 25),
    (975, 25),
    (25, 475),
    (975, 475)
]

for cx, cy in corner_points:

    draw.polygon(
        [
            (cx, cy - 6),
            (cx + 6, cy),
            (cx, cy + 6),
            (cx - 6, cy)
        ],
        fill="#B8860B"
    )


# ------------------------------------------------
# DISPLAY
# ------------------------------------------------

plt.figure(figsize=(14, 7))

plt.imshow(image)

plt.axis("off")

plt.tight_layout(pad=0)

plt.show()
```

```python
# Save final Word Cloud as JPG with a unique filename

image.save(
    "Aaradhya_Hobbies_Final_Submission.jpg",
    format="JPEG",
    quality=98,
    dpi=(300, 300)
)

print("JPG saved successfully!")
print("File: Aaradhya_Hobbies_Final_Submission.jpg")
```
