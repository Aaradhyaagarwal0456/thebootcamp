```python
# Create a list of numbers
numbers = [12, 45, 67, 24, 89, 32, 56, 71, 18, 93, 40, 25, 64, 77, 10]

print("Numbers in the list:")
print(numbers)

print("\nEven numbers in the list:")

for num in numbers:
    if num % 2 == 0:
        print(num)
```

```python
#1 Create a string
text = "Aaradhya is attending the bootcamp"

print("Original string:")
print(text)

#2 Reverse the complete string
print("\nString in reverse order:")
print(text[::-1])

#3 Reverse the order of words
words = text.split()

print("\nWords in reverse order:")
print(" ".join(words[::-1]))
```
